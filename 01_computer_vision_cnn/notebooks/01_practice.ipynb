```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
```

#### 이미지 불러오기 및 화면에 표시

```python
# PIL 라이브러리를 이용하여 이미지를 불러옵니다.
image = Image.open('../data/dog.jpg')

# 이미지를 화면에 표시합니다.
image.show()
```

```python
# matplotlib를 이용하여 이미지를 화면에 표시하는 함수 정의
def show_image(image: Image.Image, figsize: tuple[float, float]=(8, 6)) -> None:
    """
    이미지를 화면에 표시하는 함수입니다.

    Parameters:
    - image: PIL.Image 객체
    - figsize: 이미지 크기 (가로, 세로)
    """
    plt.figure(figsize=figsize)
    plt.imshow(image)
    plt.axis('off')  # 축 제거
    plt.show()

show_image(image)
```

#### 이미지를 넘파이로 변환 후 shape과 dtype 확인

```python
# 이미지를 넘파이로 변환 후 shape과 dtype 확인
img_array = np.array(image)
print(img_array.shape) # -> (높이, 너비, 채널) 순서로 출력
print(img_array.dtype) # -> uint8 (0~255 범위의 정수)
```

#### 특정 위치의 픽셀 값 확인 및 RGB 채널 분리

```python
# 특정 위치의 픽셀 값 확인 (numpy 배열은 [행(y), 열(x)] 순서로 접근)
y, x = 100, 150 # 예시로 (100, 150) 위치의 픽셀 값 확인

print(f"픽셀 위치 ({y}, {x})의 RGB 값: {img_array[y, x]}")
print(f"R: {img_array[y, x, 0]}, G: {img_array[y, x, 1]}, B: {img_array[y, x, 2]}")
```

```python
# RGB 채널 분리 (각 채널은 (높이, 너비) 크기의 2차원 배열)
r, g, b = img_array[:, :, 0], img_array[:, :, 1], img_array[:, :, 2]
print(f"R 채널 shape: {r.shape}, G 채널 shape: {g.shape}, B 채널 shape: {b.shape}")

# matplotlib를 이용하여 각 채널을 화면에 표시
fig, axes = plt.subplots(1, 3, figsize=(12, 4))
for ax, channel, name in zip(axes, [r, g, b], ['R', 'G', 'B']):
    ax.imshow(channel, cmap='gray', vmin=0, vmax=255)
    ax.set_title(f'{name} Channel')
    ax.axis('off')
plt.show()
```
